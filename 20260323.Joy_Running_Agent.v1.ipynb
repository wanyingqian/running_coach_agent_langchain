## 1. Load Packages and Environment

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage, AIMessage
from pprint import pprint
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
from IPython.display import display, Markdown


load_dotenv()
```

```python
# model initialization and testing
#model = init_chat_model(model="gpt-5-nano", temperature = 1.0, max_tokens = 1000, timeout = 200, max_retries = 1)
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature = 1.0, max_tokens = 10000, timeout = 200, max_retries = 1)
```

```python
response = model.invoke("你好你是一个精通运动康复和中长跑的专家。请你接下来为我规划我的跑步训练方案。我是一个有半年跑步经验的新手，请问您了解我们目前的任务了吗？简要回答谢谢。")
response
print(response.content)
pprint(response.response_metadata)
```

## 2. Agent

```python
# Agent initialization
agent = create_agent(model=model)

response = agent.invoke(
    {"messages": [HumanMessage(content="你好你是一个精通运动康复和中长跑的专家。请你接下来为我规划我的跑步训练方案。我是一个有半年跑步经验的新手，请问您了解我们目前的任务了吗？简要回答谢谢。"),
    AIMessage(content="是的，我已经清楚了我们目前的任务：**任务：** 为一位有半年跑步经验的新手，制定运动康复和中长跑的跑步训练方案。我完全理解了！"),
    HumanMessage(content="好的我两个月后需要去跑SF 10k，请你帮我规划一下这两个月我应该怎样训练。出一份以周为单位的运动和饮食方案。谢谢")]}
)

```

```python
reply_content = response['messages'][-1].content
display(Markdown(reply_content)) #format这个markdown的内容
```

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
